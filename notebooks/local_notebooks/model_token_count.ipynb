```python
import sys
import os
import datasets

from transformers import AutoTokenizer

# Add the project root directory to the Python path
project_root = os.path.abspath(os.path.join(os.getcwd(), '../..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
tokenizer = AutoTokenizer.from_pretrained("meta-llama/Llama-3.1-8B-Instruct")
```

```python
math_data_path = "../../dataset/processed_data/eval/base_UMWP_processed.json"
base_math_dataset = datasets.load_dataset("json", data_files=math_data_path)

math_data_path = "../../dataset/processed_data/eval/tuned_UMWP_processed.json"
tuned_math_dataset = datasets.load_dataset("json", data_files=math_data_path)

contex_data_path = "../../dataset/processed_data/eval/base_rajpurkar_squad_processed.json"
base_context_dataset = datasets.load_dataset("json", data_files=contex_data_path)

contex_data_path = "../../dataset/processed_data/eval/tuned_rajpurkar_squad_processed.json"
tuned_context_dataset = datasets.load_dataset("json", data_files=contex_data_path)
```

### Base model token count

```python
input_tokens = 0
output_tokens = 0

for i in range(len(base_math_dataset["train"])):
    input_tokens += len(tokenizer.encode(base_math_dataset["train"][i]["input"]))
    output_tokens += len(tokenizer.encode(base_math_dataset["train"][i]["responses"][0]))

print("Math QA")
print(f"Input tokens: {input_tokens}"   )
print(f"Output tokens: {output_tokens}")
print(f"Average input tokens: {input_tokens / len(base_math_dataset['train'])}")
print(f"Average output tokens: {output_tokens / len(base_math_dataset['train'])}")
```

```python
input_tokens = 0
output_tokens = 0

for i in range(len(base_context_dataset["train"])):
    input_tokens += len(tokenizer.encode(base_context_dataset["train"][i]["input"]))
    output_tokens += len(tokenizer.encode(base_context_dataset["train"][i]["responses"][0]))

print("Context QA")
print(f"Input tokens: {input_tokens}"   )
print(f"Output tokens: {output_tokens}")
print(f"Average input tokens: {input_tokens / len(base_context_dataset['train'])}")
print(f"Average output tokens: {output_tokens / len(base_context_dataset['train'])}")
```
