```python
import sys
import os
import datasets
from datasets import Dataset, concatenate_datasets
from typing import List, Dict, Any
import random
import json

from transformers import AutoTokenizer

# Add the project root directory to the Python path
project_root = os.path.abspath(os.path.join(os.getcwd(), '../..'))
if project_root not in sys.path:
    sys.path.insert(0, project_root)
```

```python
math_data_path = "../../dataset/processed_data/eval/base_UMWP_processed.json"
base_math_dataset = datasets.load_dataset("json", data_files=math_data_path)

math_data_path = "../../dataset/processed_data/eval/tuned_UMWP_processed.json"
tuned_math_dataset = datasets.load_dataset("json", data_files=math_data_path)

contex_data_path = "../../dataset/processed_data/eval/base_rajpurkar_squad_processed.json"
base_context_dataset = datasets.load_dataset("json", data_files=contex_data_path)

contex_data_path = "../../dataset/processed_data/eval/tuned_rajpurkar_squad_processed.json"
tuned_context_dataset = datasets.load_dataset("json", data_files=contex_data_path)
```

### Math QA estimation

```python
base_model_errors = 0
tuned_model_errors = 0
response_num = 0

for i in range(len(base_math_dataset["train"])):
    base_model_errors += base_math_dataset["train"][i]["wrong_response_number"]
    tuned_model_errors += tuned_math_dataset["train"][i]["wrong_response_number"]
    response_num += len(base_math_dataset["train"][i]["responses"])

print(f"Total number of responses: {response_num}")
print(f"Errors in base model: {base_model_errors}")
print(f"Errors in tuned model: {tuned_model_errors}")
print(f"Base model error rate: {base_model_errors / response_num}")
print(f"Tuned model error rate: {tuned_model_errors / response_num}")
```
